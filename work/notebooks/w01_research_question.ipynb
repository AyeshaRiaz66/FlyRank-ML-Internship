## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
# The question: decision, action, cost of a wrong call

## Research Question

Can we identify pages that are likely to receive lower search performance using the available search data?

## Decision

Which pages should the content or SEO team improve first?

## Action

The team can update page titles, improve content quality, add internal links, or optimize keywords for pages identified as needing attention.

## Cost of a Wrong Recommendation

If the recommendation is incorrect, the team may spend time improving pages that already perform well while ignoring pages that actually need improvement. This wastes time and resources but does not permanently harm the website.
```
