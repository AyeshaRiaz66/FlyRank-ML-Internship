## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
# My lane as an ML task

## Lane
Content Performance Prediction

## ML Task Type
Scoring (Regression)

## Why

The goal is to estimate how well a content page is likely to perform using historical data. Instead of assigning pages to simple categories, the model produces a score or predicted performance value that helps prioritize which pages should be reviewed first.
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
# Target or Proxy

The ideal target is future content performance after a page is updated. Since future data is not available, a proxy target can be created from the historical dataset, such as the current performance score, click count, traffic level, or another performance-related column available in the data.

The target should represent how successful a page is so that similar pages can be prioritized for improvement.
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# Success Metric

The success of this project is measured by how well the predicted scores match the observed values in the data. If the project is treated as a regression problem, metrics such as Mean Absolute Error (MAE) or Root Mean Squared Error (RMSE) are appropriate.

From a business perspective, success means helping the content team identify pages that deserve attention before spending time updating every page.
```

*What makes the pattern too messy for an if-statement?*

```python
# Why ML beats a fixed rule

A fixed rule might say that pages with low traffic should always be updated first. However, page performance depends on multiple factors that interact with each other, such as historical traffic, content characteristics, engagement, and other available features.

Machine learning can learn these patterns from historical data and provide a more informed ranking than a single manually defined rule. The output supports decision-making but does not replace human judgment.
```
