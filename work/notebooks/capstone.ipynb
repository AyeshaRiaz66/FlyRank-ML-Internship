## 1. Question

*The research question and the decision it supports.*

```python
Can search performance signals predict which pages are at risk of a meaningful traffic decline, well enough to prioritize a content team's refresh queue?
```

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

```python
FlyRank Internship Warehouse (Hugging Face, gated release), fact_content_daily and fact_content_query_90d tables, aggregated over the full ~79M-row release via DuckDB. Date window: [fill in from your bounds query — MIN/MAX report_date]. Excluded: content items with fewer than 150 impressions in the prior 90-day window (too little history for a stable signal). No client names, domains, URLs, or raw query text included — anonymized hashes only.
```

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

```python
Label: impressions in the last 90-day window down >20% vs. the prior 90-day window. Features computed only from the prior window (imp_prev90, rare-query share, visible query count, anonymized-impression share, top-query concentration, weekend click share) to avoid leakage. Baseline: Week 4 rule-based score, reconstructed with warehouse-available features only, bottom-35th-percentile threshold. Validation: random 75/25 split vs. GroupShuffleSplit on client_hash_id (whole clients held out) to test cross-client generalization.
```

## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

```python
	Accuracy	F1 (declining)
Random Forest — random split	0.864	0.844
Random Forest — group split (unseen clients)	0.849	0.845
Rule baseline (same features)	0.465	0.360

Model beats baseline by ~38 points of accuracy on the same held-out clients. Only a 1.5-point drop from random to group split — signal generalizes, isn't client-memorization.
```

## 5. Limitations

*What this work cannot claim.*

```python
Label is a proxy for "at risk," not confirmed business harm — some declines are seasonal/expected. Baseline is a simplified reconstruction, missing CTR/engagement/content-age used in the original rule, so the comparison isn't fully apples-to-apples. Top feature (imp_prev90) is mechanically related to how the label itself is computed — a real signal, but flagged for further scrutiny, not treated as causal. Results are observed associations in this dataset/window, not a claim about what refreshing any specific page will do.
```

## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

```python
Rank flagged pages by predicted decline probability. High risk + low query diversity → broaden query targeting / improve SEO. High risk + high rare-query share → content may be too niche, expand coverage. Borderline predictions → manual review before action, not automatic refresh.
```
