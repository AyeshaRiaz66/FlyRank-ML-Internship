## 1. Unit of analysis + time window

*One row = one what, over which dates? State it, then verify it below.*

```python
## Unit of Analysis

One row represents the daily search performance of one content page.

## Time Window

For exploration, I will use a mid-panel month (2026-03). This avoids using the final month as training data and keeps the final month available as a future evaluation period.
```

```python
SELECT
    COUNT(*) AS total_rows,
    MIN(date) AS first_date,
    MAX(date) AS last_date
FROM fact_content_daily_performance
WHERE month='2026-03';
```
